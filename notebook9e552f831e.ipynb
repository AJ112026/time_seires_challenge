# HMS – Harmful Brain Activity Classification: a simple scikit-learn baseline

**Task.** Each sample is an EEG segment from a critically ill patient. Several experts voted on which of six
patterns it shows: seizure, LPD, GPD, LRDA, GRDA or other. The target is the *distribution* of votes,
and submissions are scored with the KL divergence between the true and the predicted distribution (lower is better).

**Approach.**
1. **Labels**: for each EEG, turn the expert votes into a probability distribution over the six classes.
2. **Features** (88 per EEG): from the official Kaggle spectrograms we take the 10 minutes around the labelled
   segment and compute log band power (delta, theta, alpha, beta) for each of the four scalp regions, summarised
   over time (mean, std, min, max), plus relative band power and left–right asymmetry.
3. **Models**: `LogisticRegression` and `HistGradientBoostingClassifier` from scikit-learn.
   Soft labels are handled by repeating each EEG once per class with `sample_weight` = vote fraction;
   the weighted cross-entropy is then equal to the KL divergence up to a constant.
4. **Validation**: 5-fold `GroupKFold` grouped by `patient_id`, so no patient appears in both the training and
   the validation part of a fold. The model with the better cross-validation (CV) score is used for the submission.
5. **Submission**: refit the chosen model on all training data and predict the hidden test set.

**Settings** (next cell): `MODE = "full"` uses all training spectrograms (about 10 minutes on a Kaggle CPU);
`MODE = "quick"` uses a random 3,000 for a fast check. `RUN_CV = False` skips cross-validation.

Runs on CPU only; internet must be off for Kaggle submissions.

```python
import os
import time
import warnings
from glob import glob

import numpy as np
import pandas as pd

from joblib import Parallel, delayed
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GroupKFold
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# Some spectrograms contain frequency bands that are entirely NaN; numpy warns about the
# resulting empty means. These cases are handled explicitly below, so the warnings are muted.
warnings.filterwarnings("ignore", category=RuntimeWarning)

# ----------------------------------------------------------------------------- settings
MODE = "full"      # "full": all training spectrograms | "quick": a random subset for a fast check
RUN_CV = True     # True: run 5-fold cross-validation, report CV scores and pick the better model
MAX_SPECS = {"full": None, "quick": 3000}[MODE]   # number of training spectrograms to read (None = all)
N_FOLDS = 5        # number of cross-validation folds
N_JOBS = -1        # number of CPU cores used to read spectrograms in parallel (-1 = all cores)
SEED = 42          # random seed for subsampling and for the gradient-boosting model

# ----------------------------------------------------------------------------- data location
# Kaggle mounts the competition data at different paths depending on the notebook environment.
# HMS_DATA can be set to run the notebook elsewhere (e.g. on a local copy of the data).
CANDIDATES = ["/kaggle/input/competitions/hms-harmful-brain-activity-classification",
              "/kaggle/input/hms-harmful-brain-activity-classification"]
DATA = os.environ.get("HMS_DATA") or next((p for p in CANDIDATES if os.path.exists(f"{p}/train.csv")), None)
if DATA is None:   # last resort: search the input folder (slow, about 1-2 minutes)
    DATA = os.path.dirname(sorted(glob("/kaggle/input/**/train.csv", recursive=True), key=len)[0])
print("DATA =", DATA)

# ----------------------------------------------------------------------------- constants
# Vote columns in train.csv, in the order required by the submission file.
CLASSES = ["seizure_vote", "lpd_vote", "gpd_vote", "lrda_vote", "grda_vote", "other_vote"]
# The official spectrograms average the bipolar EEG channels into four scalp regions:
# LL = left lateral, RL = right lateral, LP = left parasagittal, RP = right parasagittal.
REGIONS = ["LL", "RL", "LP", "RP"]
# Classical EEG frequency bands in Hz (the spectrograms cover roughly 0.5-20 Hz).
BANDS = {"delta": (0.5, 4), "theta": (4, 8), "alpha": (8, 13), "beta": (13, 20)}
```

## 1. Labels: one row per EEG, votes normalised to a probability distribution

`train.csv` has several rows (labelled sub-segments) per `eeg_id`. We sum the votes over these rows,
divide by the total number of votes, and keep the range of label offsets to locate the window in the spectrogram.

```python
train = pd.read_csv(f"{DATA}/train.csv")

# One row per EEG: spectrogram id, first/last label offset (seconds), patient id and summed votes per class.
agg = {"spectrogram_id": "first", "spectrogram_label_offset_seconds": ["min", "max"], "patient_id": "first"}
agg.update({c: "sum" for c in CLASSES})
df = train.groupby("eeg_id").agg(agg)
df.columns = ["spectrogram_id", "off_min", "off_max", "patient_id"] + CLASSES

# Convert vote counts into a probability distribution (each row sums to 1).
df["num_votes"] = df[CLASSES].sum(axis=1)
df[CLASSES] = df[CLASSES].values / df["num_votes"].values[:, None]
df = df.reset_index()

# Quick mode: keep only the EEGs of a random subset of spectrograms.
if MAX_SPECS:
    keep = pd.Series(df.spectrogram_id.unique()).sample(n=min(MAX_SPECS, df.spectrogram_id.nunique()),
                                                         random_state=SEED)
    df = df[df.spectrogram_id.isin(keep)].reset_index(drop=True)

print(df.shape, "EEGs from", df.patient_id.nunique(), "patients,", df.spectrogram_id.nunique(), "spectrograms")
df.head()
```

## 2. Features from the official spectrograms

Each spectrogram file has one row every 2 seconds and 400 power columns named `<region>_<frequency>`
(e.g. `LL_1.17`). We take a 10-minute window (300 rows) and reduce it to 88 numbers per EEG:

| Feature group | Count |
|---|---|
| Log band power per region and band: mean, std, min, max over time | 4 regions × 4 bands × 4 = 64 |
| Relative band power (band power / total power of the region) | 4 × 4 = 16 |
| Left–right asymmetry of mean log band power (lateral and parasagittal chains) | 2 × 4 = 8 |

Spectrogram files are read in parallel because reading them is the slowest step of the notebook.

```python
def band_columns(columns):
    """Return {(region, band): [column indices]} for the 400 spectrogram columns like 'LL_1.17'."""
    out = {}
    for r in REGIONS:
        for b, (lo, hi) in BANDS.items():
            out[(r, b)] = [i for i, c in enumerate(columns)
                           if c.startswith(r + "_") and lo <= float(c.split("_")[1]) < hi]
    return out


def window_features(spec, cols, prefix=""):
    """Compute the 88 features for one time window."""
    feats = {}
    if len(spec) == 0:
        return feats
    total = {r: 0.0 for r in REGIONS}   # total power per region, for relative band power
    band_power = {}
    for (r, b), idx in cols.items():
        p = np.nanmean(spec[:, idx], axis=1)                     # mean power in the band, per time step
        lp = np.log1p(np.nan_to_num(p, nan=0.0).clip(min=0))     # log scale; missing values -> 0
        feats[f"{prefix}{r}_{b}_mean"] = lp.mean()
        feats[f"{prefix}{r}_{b}_std"] = lp.std()
        feats[f"{prefix}{r}_{b}_min"] = lp.min()
        feats[f"{prefix}{r}_{b}_max"] = lp.max()
        band_power[(r, b)] = np.nanmean(p) if np.isfinite(p).any() else 0.0
        total[r] += band_power[(r, b)]
    for (r, b), v in band_power.items():                         # share of the region's total power
        feats[f"{prefix}{r}_{b}_rel"] = v / (total[r] + 1e-6)
    for b in BANDS:                                              # left minus right hemisphere
        feats[f"{prefix}asym_lat_{b}"] = feats[f"{prefix}LL_{b}_mean"] - feats[f"{prefix}RL_{b}_mean"]
        feats[f"{prefix}asym_par_{b}"] = feats[f"{prefix}LP_{b}_mean"] - feats[f"{prefix}RP_{b}_mean"]
    return feats
#die window features zijn aangepast, prefix meegegeven zodat dezelfde 88 spectrogram-statistieken zonder naamconflicten op meerdere tijdschalen berekend kunnen worden.

def spectrogram_features(path, windows, is_test):
    """Read one spectrogram file and compute multi-window features for every EEG that uses it."""
    spec = pd.read_parquet(path)
    cols = band_columns(list(spec.columns[1:]))
    spec = spec.iloc[:, 1:].to_numpy(dtype=np.float32)          # drop the 'time' column
    rows = []
    for eeg_id, off_min, off_max in windows:
        if is_test:
            start = max(0, (len(spec) - 300) // 2)
        else:
            start = int((off_min + off_max) // 4)

        win_10m = spec[start:start + 300]                        # 10 minuten (300 rijen)
        mid = len(win_10m) // 2
        win_3m = win_10m[max(0, mid - 45):mid + 45]              # middelste 3 minuten (90 rijen)
        win_50s = win_10m[max(0, mid - 12):mid + 13]             # middelste 50 seconden (25 rijen)

        feats = {}
        feats.update(window_features(win_10m, cols, prefix="w10m_"))
        feats.update(window_features(win_3m, cols, prefix="w3m_"))
        feats.update(window_features(win_50s, cols, prefix="w50s_"))

        # Piek in het midden (50s) ten opzichte van de hele 10 minuten
        for r in REGIONS:
            for b in BANDS:
                k_50s = f"w50s_{r}_{b}_mean"
                k_10m = f"w10m_{r}_{b}_mean"
                if k_50s in feats and k_10m in feats:
                    feats[f"diff_50s_10m_{r}_{b}"] = feats[k_50s] - feats[k_10m]

        feats["eeg_id"] = eeg_id
        rows.append(feats)
    return rows

#OPM: in plaats van gemiddeldes over de volle 10 min te pakken moet je die middenste 50s gebruiken, met rede dat hier
#enkel de dokters over geredeneerd hebben. De code trekt het gemiddelde voor elke hersenratio van de 10min af van die van de 50s, en dan 
#ziet men of er een verschil is of niet
#Het 10-minutenvenster (300 rijen) opgesplitst in drie tijdschalen: 
#de volledige 10 minuten (achtergrondcontext), de middelste 3 minuten (90 rijen) en de middelste 50 seconden (25 rijen, het exacte beoordelingsvenster van de neurologen). Daarnaast 16 contrast-features (w50s - w10m) toegevoegd om plotselinge pieken in hersenactiviteit te vangen (totaal: 280 features).






def build_features(meta, spec_dir, is_test=False):
    """Compute features for all EEGs in `meta`, reading each spectrogram file once (in parallel)."""
    t0 = time.time()
    if is_test:   # test.csv has no label offsets
        meta = meta.assign(off_min=0.0, off_max=0.0)
    jobs = [delayed(spectrogram_features)(f"{spec_dir}/{sid}.parquet",
                                          list(zip(g.eeg_id, g.off_min, g.off_max)), is_test)
            for sid, g in meta.groupby("spectrogram_id")]
    results = Parallel(n_jobs=N_JOBS, batch_size=32)(jobs)
    out = pd.DataFrame([row for rows in results for row in rows])
    print(f"{len(jobs)} spectrograms -> {len(out)} feature rows in {time.time() - t0:.0f}s")
    return out


X_df = build_features(df, f"{DATA}/train_spectrograms")
df = df.merge(X_df, on="eeg_id")
FEATS = [c for c in X_df.columns if c != "eeg_id"]
print(len(FEATS), "features")
```

```python
#Nieuwe cel die die ruwe EEG's gaat invoeren
# ----------------------------------------------------------------------------- ruwe EEG-features (NIEUW)
from scipy.signal import welch, butter, sosfiltfilt

FS = 200                     # de ruwe EEG heeft 200 metingen per seconde
WIN = 50 * FS                # 50 seconden = het stuk dat de experts beoordeelden
# "Double banana"-montage: 4 ketens van 4 bipolaire kanalen (verschil tussen twee naburige elektroden).
# Dit zijn dezelfde 4 zones als in de spectrogrammen (LL, RL, LP, RP).
CHAINS = {
    "LL": [("Fp1", "F7"), ("F7", "T3"), ("T3", "T5"), ("T5", "O1")],
    "RL": [("Fp2", "F8"), ("F8", "T4"), ("T4", "T6"), ("T6", "O2")],
    "LP": [("Fp1", "F3"), ("F3", "C3"), ("C3", "P3"), ("P3", "O1")],
    "RP": [("Fp2", "F4"), ("F4", "C4"), ("C4", "P4"), ("P4", "O2")],
}
ELECTRODES = sorted({e for pairs in CHAINS.values() for pair in pairs for e in pair})
SOS = butter(4, [0.5, 20], btype="band", fs=FS, output="sos")   # bandpass-filter 0.5-20 Hz


def eeg_window_features(sig, prefix):
    """Band power, std en line length per keten (gemiddeld over de 4 kanalen van de keten)."""
    feats = {}
    for ch, x in sig.items():                                    # x: (4 kanalen, tijd)
        f, psd = welch(x, fs=FS, nperseg=min(2 * FS, x.shape[1]), axis=1)
        bp = {b: psd[:, (f >= lo) & (f < hi)].mean(axis=1) for b, (lo, hi) in BANDS.items()}
        total = sum(bp.values()) + 1e-6
        for b in BANDS:
            feats[f"{prefix}{ch}_{b}_logpow"] = np.log1p(bp[b]).mean()   # hoeveel activiteit in de band
            feats[f"{prefix}{ch}_{b}_rel"] = (bp[b] / total).mean()      # aandeel van de band
        feats[f"{prefix}{ch}_logstd"] = np.log1p(x.std(axis=1)).mean()  # grootte van het signaal
        # line length: hoeveel het signaal op en neer gaat (hoog bij seizures)
        feats[f"{prefix}{ch}_linelen"] = np.log1p(np.abs(np.diff(x, axis=1)).mean(axis=1)).mean()
    for b in BANDS:                                              # links min rechts
        feats[f"{prefix}asym_lat_{b}"] = feats[f"{prefix}LL_{b}_logpow"] - feats[f"{prefix}RL_{b}_logpow"]
        feats[f"{prefix}asym_par_{b}"] = feats[f"{prefix}LP_{b}_logpow"] - feats[f"{prefix}RP_{b}_logpow"]
    return feats


def envelope(x, bin_size=FS // 10):
    """Gemiddelde amplitude per 0.1 s, gemiddeld over de 4 kanalen van een keten."""
    n = x.shape[1] // bin_size
    return np.abs(x[:, :n * bin_size]).reshape(x.shape[0], n, bin_size).mean(axis=(0, 2))


def periodicity(env, min_lag=5, max_lag=30):
    """Hoe sterk herhaalt het signaal zich om de 0.5-3 s (typisch voor LPD/GPD)? -> (sterkte, periode in s)."""
    e = env - env.mean()
    denom = (e * e).sum()
    if denom == 0:
        return 0.0, 0.0
    ac = np.array([(e[:-k] * e[k:]).sum() / denom for k in range(min_lag, max_lag + 1)])
    i = int(ac.argmax())
    return ac[i], (min_lag + i) / 10


def eeg_features(path, eeg_id, offset):
    """Lees één ruwe EEG en bereken de features voor het 50 s-venster vanaf `offset` (seconden)."""
    eeg = pd.read_parquet(path, columns=ELECTRODES)
    start = int(offset * FS)
    if start + WIN > len(eeg):                                   # venster mag niet voorbij het einde
        start = max(0, len(eeg) - WIN)
    w = eeg.iloc[start:start + WIN]
    w = w.fillna(w.mean()).fillna(0.0)                           # ontbrekende waarden opvullen
    raw = w.to_numpy(dtype=np.float64)
    if len(raw) < WIN:                                           # te kort: aanvullen met nullen
        raw = np.pad(raw, ((0, WIN - len(raw)), (0, 0)))
    col = {e: i for i, e in enumerate(ELECTRODES)}
    sig50 = {}
    for ch, pairs in CHAINS.items():
        x = np.stack([raw[:, col[a]] - raw[:, col[b]] for a, b in pairs])   # 4 bipolaire kanalen
        sig50[ch] = sosfiltfilt(SOS, np.clip(x, -1000, 1000), axis=1)
    sig10 = {ch: x[:, 20 * FS:30 * FS] for ch, x in sig50.items()}       # middelste 10 s (het label)

    feats = {"eeg_id": eeg_id}
    feats.update(eeg_window_features(sig50, "e50_"))
    feats.update(eeg_window_features(sig10, "e10_"))
    env = {ch: envelope(x) for ch, x in sig50.items()}
    for ch, e in env.items():
        feats[f"e50_{ch}_period_strength"], feats[f"e50_{ch}_period_sec"] = periodicity(e)
    # samenhang links-rechts: hoog bij gegeneraliseerde (GPD/GRDA), laag bij gelateraliseerde (LPD/LRDA)
    feats["e50_corr_lat"] = np.corrcoef(env["LL"], env["RL"])[0, 1]
    feats["e50_corr_par"] = np.corrcoef(env["LP"], env["RP"])[0, 1]
    return feats


def build_eeg_features(meta, eeg_dir, is_test=False):
    """Bereken de ruwe EEG-features voor alle EEG's in `meta` (parallel)."""
    t0 = time.time()
    offsets = np.zeros(len(meta)) if is_test else meta["eeg_off"].to_numpy()   # test-EEG's zijn precies 50 s
    rows = Parallel(n_jobs=N_JOBS, batch_size=16)(
        delayed(eeg_features)(f"{eeg_dir}/{eid}.parquet", eid, off)
        for eid, off in zip(meta["eeg_id"], offsets))
    out = pd.DataFrame(rows)
    print(f"{len(out)} EEG's -> {out.shape[1] - 1} EEG-features in {time.time() - t0:.0f}s")
    return out


# Startpunt van het 50 s-venster: midden van de gelabelde offsets van die EEG (zelfde idee als bij het spectrogram).
eeg_off = train.groupby("eeg_id")["eeg_label_offset_seconds"].agg(["min", "max"]).mean(axis=1)
df["eeg_off"] = df["eeg_id"].map(eeg_off)

E_df = build_eeg_features(df, f"{DATA}/train_eegs")
df = df.merge(E_df, on="eeg_id")
FEATS = FEATS + [c for c in E_df.columns if c != "eeg_id"]
print(len(FEATS), "features in totaal (spectrogram + ruwe EEG)")
```

## 3. Soft-label training and patient-grouped cross-validation

**Soft labels in scikit-learn.** scikit-learn classifiers expect one class per sample. We therefore repeat each EEG
six times, once per class, and weight each copy by that class's vote fraction. Minimising the weighted
cross-entropy is then the same as minimising the KL divergence to the vote distribution.

**Cross-validation (CV).** The training data are split into 5 folds *by patient*. Each model is trained on 4 folds
and evaluated on the held-out fold; the out-of-fold predictions give an estimate of performance on unseen patients.
Grouping by patient matters: EEGs of the same patient are very similar, and a random split would overestimate
performance. We report the CV score on all EEGs and on EEGs with at least 10 votes (more reliable labels),
next to the score of a trivial model that always predicts the average vote distribution (the *prior*).

```python
import itertools
from sklearn.ensemble import ExtraTreesClassifier


def kl_score(y_true, y_pred, eps=1e-15):
    """Mean KL divergence KL(y_true || y_pred) over samples, as in the competition metric."""
    p = np.clip(y_pred, eps, 1)
    p = p / p.sum(axis=1, keepdims=True)
    t = y_true
    return np.mean(np.sum(np.where(t > 0, t * np.log(np.where(t > 0, t, 1) / p), 0), axis=1))


def expand(X, Y, w=None):
    """Soft labels -> gewogen harde labels. Elke EEG wordt per klasse herhaald.
    Gewicht = stemfractie (x betrouwbaarheid van de EEG als w gegeven is)."""
    n, k = Y.shape
    Xr = np.repeat(X, k, axis=0)
    yr = np.tile(np.arange(k), n)
    wr = Y.reshape(-1) if w is None else (Y * w[:, None]).reshape(-1)   # <- aangepast in versie 3, met gewicht 
    #geven aan kopie, nu wordt er rekeninggehouden met de betrouwbaarheid (#stemmen) van de EEG
    keep = wr > 0
    return Xr[keep], yr[keep], wr[keep]


MODELS = {
    "logreg": lambda: make_pipeline(StandardScaler(), LogisticRegression(C=0.1, max_iter=2000)),
    "hgb": lambda: HistGradientBoostingClassifier(learning_rate=0.05, max_iter=600,
                                                  max_leaf_nodes=31, min_samples_leaf=50,
                                                  l2_regularization=1.0, max_features=0.5,
                                                  random_state=SEED),
}
#ervoor stond hier nog het et-model met uitleg: 
# Vierde aanpassing: ExtraTrees = veel losse, willekeurige bomen waarvan het gemiddelde genomen wordt.
    # min_samples_leaf=20: elk blad bevat minstens 20 rijen, zodat de kansen niet te extreem worden.
def fit(name, X, Y, w=None):
    """Train model `name` op features X, soft labels Y en (optioneel) gewicht per EEG w."""
    Xe, ye, we = expand(X, Y, w)
    m = MODELS[name]()
    if name == "logreg":   # in een Pipeline moet het gewicht naar de laatste stap
        m.fit(Xe, ye, logisticregression__sample_weight=we)
    else:
        m.fit(Xe, ye, sample_weight=we)
    return m


def blend(preds, weights, pri, b):
    """Hybride voorspelling: gewogen gemiddelde van alle modellen, gemengd met een aandeel b prior.
    preds = {model: voorspellingen}, weights = {model: gewicht} (gewichten tellen op tot 1)."""
    p = sum(weights[n] * preds[n] for n in weights)
    return (1 - b) * p + b * pri


X = np.nan_to_num(df[FEATS].to_numpy(dtype=np.float32), nan=0.0, posinf=0.0, neginf=0.0)
Y = df[CLASSES].to_numpy()
groups = df["patient_id"].to_numpy()
prior = np.tile(Y.mean(axis=0), (len(Y), 1))            # dom referentiemodel: altijd de gemiddelde verdeling
hi = df["num_votes"].to_numpy() >= 10                    # EEG's met 10+ stemmen (voor de 2de score-regel)
W = np.minimum(df["num_votes"].to_numpy(), 10) / 10      # gewicht per EEG: 1 stem -> 0.1, 10+ stemmen -> 1

if RUN_CV:
    oof = {name: np.zeros_like(Y) for name in MODELS}    # out-of-fold voorspellingen per model
    for fold, (tr, va) in enumerate(GroupKFold(n_splits=N_FOLDS).split(X, Y, groups)):
        for name in MODELS:
            oof[name][va] = fit(name, X[tr], Y[tr], W[tr]).predict_proba(X[va])
        print(f"fold {fold}: " + ", ".join(f"{n}={kl_score(Y[va], oof[n][va]):.4f}" for n in MODELS))
    print(f"\nCV KL (all EEGs)        prior={kl_score(Y, prior):.4f} " +
          " ".join(f"{n}={kl_score(Y, oof[n]):.4f}" for n in MODELS))
    print(f"CV KL (>=10 votes only) prior={kl_score(Y[hi], prior[hi]):.4f} " +
          " ".join(f"{n}={kl_score(Y[hi], oof[n][hi]):.4f}" for n in MODELS))

    # Hybride model: probeer alle mixen van de modellen (stappen van 0.1, samen 1) en een aandeel prior,
    # en kies de mix met de laagste KL op de EEG's met 10+ stemmen.
    names = list(MODELS)
    steps = np.round(np.arange(0, 1.01, 0.1), 1)
    mixes = [dict(zip(names, c)) for c in itertools.product(steps, repeat=len(names)) if abs(sum(c) - 1) < 1e-6]
    oof_hi = {n: oof[n][hi] for n in names}
    WEIGHTS, B = min(((m, b) for m in mixes for b in [0, 0.05, 0.1, 0.15, 0.2]),
                     key=lambda mb: kl_score(Y[hi], blend(oof_hi, mb[0], prior[hi], mb[1])))
    oof_blend = blend(oof, WEIGHTS, prior, B)
    print("\nBlend: " + " ".join(f"{n}={w:.1f}" for n, w in WEIGHTS.items()) + f" prior={B:.2f}")
    print(f"CV KL (all EEGs)        blend={kl_score(Y, oof_blend):.4f}")
    print(f"CV KL (>=10 votes only) blend={kl_score(Y[hi], oof_blend[hi]):.4f}")
else:
    WEIGHTS, B = {"hgb": 1.0}, 0.0   # zonder CV: gewoon 100 % hgb
```

## 4. Fit the chosen model on all training data and predict the test set

The visible `test.csv` contains a single example. When the notebook is submitted, Kaggle re-runs it
